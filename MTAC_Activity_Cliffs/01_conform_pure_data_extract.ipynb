```python
# Imports
import pandas as pd
import numpy as np
from rdkit import Chem
from rdkit.Chem import AllChem
```

```python
df = pd.read_csv("CHEMBL203_raw_data.csv", sep=';')
df
```

```python
# Remove extra spaces in column names (important)
df.columns = df.columns.str.strip()

# Count IC50
ic50_count = (df["Standard Type"] == "IC50").sum()

# Count Ki
ki_count = (df["Standard Type"] == "Ki").sum()

print("Number of IC50 values:", ic50_count)
print("Number of Ki values:", ki_count)
```

```python
df[df["Standard Type"] == "Ki"]["Standard Relation"].value_counts()
''''
Ki = 50 nM, This means the measured inhibition constant is exactly 50 nM.
These are good for ML modeling.
Ki > 1000 nM
The compound was tested, but inhibition was weaker than 1000 nM.
These are called censored data
ML regression needs exact numeric values.
> and < introduce noise.
Activity cliff analysis requires precise Δactivity.
> and < values are called:
Right-censored data
Left-censored data
'''''

```
